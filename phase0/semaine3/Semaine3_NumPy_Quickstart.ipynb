# NumPy Quickstart : Les fondamentaux pour la Data Science

NumPy est la bibliothèque fondamentale pour le calcul scientifique. Elle fournit l'objet `ndarray` pour des opérations vectorisées efficaces.

## 1. Les bases de l'objet ndarray
Un array possède des attributs essentiels :
- **ndim** : le nombre d'axes (dimensions).
- **shape** : la taille du tableau (ex: `(3, 5)`).
- **size** : le nombre total d'éléments.
- **dtype** : le type des éléments (int64, float64, etc.).


### Exemple d'attributs
import numpy as np

a = np.arange(15).reshape(3, 5)
print(f"Array:\n{a}")
print(f"Shape: {a.shape}, Dimensions: {a.ndim}, Type: {a.dtype}, Taille: {a.size}")


## Commencer par lancer la bibliothèque numpy

```python
import numpy as np
```

## 2. Création de Tableaux

```python
# Création à partir de listes
a = np.array([2, 3, 4]) 
b = np.array([(1.5, 2, 3), (4, 5, 6)]) # 2D array

# Tableaux remplis de valeurs par défaut (très utile pour initialiser des poids de réseaux de neurones)
zeros = np.zeros((3, 4))
ones = np.ones((2, 3, 4), dtype=np.int16)
empty = np.empty((2, 3)) # Mémoire brute, contenu aléatoire (plus rapide)

# Séquences numériques
print("Arange (pas):", np.arange(0, 2, 0.3)) 
print("Linspace (nombre d'éléments):", np.linspace(0, 2, 9)) # Recommandé pour les nombres flottants

# Spécification du type complexe
c = np.array([[1, 2], [3, 4]], dtype=complex)
```

## 3. Opérations de Base
Les opérations sont par défaut **élément par élément** (element-wise).


```python
a = np.array([20, 30, 40, 50])
b = np.arange(4)

print("Soustraction:", a - b)
print("Puissance:", b**2)
print("Masque booléen (très utilisé en Pandas):", a < 35)

# Multiplication matricielle
A = np.array([[1, 1], [0, 1]])
B = np.array([[2, 0], [3, 4]])

print("Produit élément par élément (*): \n", A * B)
print("Produit matriciel (@): \n", A @ B) # Ou A.dot(B)
```

## 4. Indexation, Slicing et Itération
Fonctionne comme les listes Python, mais sur plusieurs dimensions.


```python
def f(x, y):
    return 10 * x + y

b = np.fromfunction(f, (5, 4), dtype=int)
print("Array complet:\n", b)
print("Deuxième ligne, troisième colonne:", b[2, 3])
print("Colonnes 1 à 3 de chaque ligne:\n", b[0:5, 1]) # Equivalent b[:, 1]
```

## 5. Manipulation de Forme (Shape)
C'est ici que se jouent les performances en ML (Deep Learning).


```python
a = np.floor(10 * np.random.random((3, 4)))
print("Original:\n", a)
print("Aplatir (flatten):", a.ravel())
print("Redimensionner (3x4 -> 6x2):\n", a.reshape(6, 2))
print("Transposée:\n", a.T)

# Astuce : reshape(-1) calcule automatiquement la dimension manquante
print("Automatique (un seul axe):\n", a.reshape(2, -1))
```

## 6. Empilement (Stacking) et Découpage
Essentiel pour fusionner des jeux de données.


```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print("Vertical stack:\n", np.vstack((a, b)))
print("Horizontal stack:\n", np.hstack((a, b)))
```

## 7. Copies et Vues : Le piège à éviter
C'est la partie la plus importante pour éviter les bugs silencieux.


```python
# 1. Simple assignation (Pas de copie du tout)
a = np.arange(12)
b = a 
print("b est a ?", b is a) # True

# 2. Vue ou "Shallow Copy" (Partage les données mais pas la forme)
c = a.view()
print("c est a ?", c is a) # False
print("c partage les données de a ?", c.base is a) # True
c.shape = 2, 6
a[0] = 1234
print("a a changé, c aussi :", c[0, 0]) # c est modifié !

# 3. Copie profonde (Deep Copy)
d = a.copy()
d[0] = -1
print("a reste inchangé :", a[0]) # d est indépendant
```

## 8. Fonctions Universelles (ufunc)
NumPy fournit des fonctions mathématiques rapides qui s'appliquent sur tout le tableau.


```python
B = np.arange(3)
print("Exponentielle:", np.exp(B))
print("Racine carrée:", np.sqrt(B))
C = np.array([2., -1., 4.])
print("Somme:", np.add(B, C))
```

### Résumé pour la suite
1. **Vectorisez** vos calculs : ne faites jamais de boucles `for` sur un array.
2. Vérifiez toujours le **shape** avec `a.shape`.
3. Utilisez `.copy()` si vous voulez modifier un sous-ensemble de données sans altérer l'original.
4. Le **broadcasting** permet d'opérer sur des arrays de tailles différentes (ex: ajouter un scalaire à une matrice).
