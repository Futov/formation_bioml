## Matrices

```python
data = np.array([
    [145, 6.8, 28.4],
    [118, 4.9, 22.1],
    [168, 11.3, 31.7]
])

print(data.shape)
print(data[2, 1])
print(data[0])
print(data[:,0])
```

## La transposition : retourner une matrice

```python
A = np.array([[1, 2, 3], [4, 5, 6]])

A.T
A.T.shape
data = np.array([[145, 6.8], [118, 4.9], [168, 11.3]])
data.T
```

### Broadcasting


```python
means = np.mean(data, axis=0)
print(means)
data_centree = data - means
print(data_centree)
```

### Reshape

```python
v = np.arange(12)
print(v)
m = v.reshape(3, 4)
print(m)
M3 = v.reshape(12, 1)
print(M3)
M4 = v.reshape(-1, 4)
print(M4)
```
