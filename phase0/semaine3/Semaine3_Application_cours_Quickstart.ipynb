## Création de tableau

```python
# lancement de Numpy
import numpy as np

a = np.arange(15).reshape(3, 5)
print(f"Array :\n{a}")
print(f"Shape : {a.shape}, Dimensions : {a.ndim}D, Type : {a.dtype}, Taille : {a.size}")
print(f"Poids d'un élément en mémoire : {a.itemsize} octets" )
```

```python
# Création à partir de liste
a = np.array([2, 3, 4])
# 2D : Une liste contenant des tuples (séquence de séquences)
b = np.array([(1.5, 2, 3), (4, 5, 6)])
print(f"Shape de b : {b.shape}") # (2, 3)

# 3D : Une liste contenant une liste qui contient des listes
c = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(f"Shape de c : {c.shape}") # (2, 2, 2)
print(f"Dimensions de c : {c.ndim}D")
```

```python
# 1. Zeros : Idéal pour initialiser les 'biases' en Deep Learning ou servir de compteur
zeros = np.zeros((3, 4))
print(zeros)

# 2. Ones : Utile pour les masques de sélection ou les transformations multiplicatives
ones = np.ones((2, 3, 4), dtype=np.int16)
print(f"\n{ones}")

# 3. Empty : Performance maximale. Attention, contient des données aléatoires (mémoire brute)
# À utiliser uniquement si vous allez écraser toutes les valeurs immédiatement après !
#empty = np.empty((2, 3))
#print(f'\n{empty}')

# 4. Full : Pour une valeur constante spécifique (ex: remplir avec -1 ou pi)
constant = np.full((2, 2), 3.14)
print(f'\n{constant}')
```

```python
# Séquences numériques
# arange : on définit le DEBUT, la FIN (exclue) et le PAS.
print("Arange (quand on connaît le pas) :", np.arange(0, 2, 0.3))

# linspace : on définit le DEBUT, la FIN (incluse) et le NOMBRE D'ÉLÉMENTS.
# Très utilisé pour créer des axes X pour des graphiques.
print("Linspace (nombre d'éléments) :", np.linspace(0, 2, 9))

# Spécification du type complexe (pour le traitement du signal, Fourier, etc.)
c = np.array([[1, 2], [3, 4]], dtype=complex)
print(f"\nTableau complexe :\n{c}")
```

## Opération de base

```python
a = np.array([20, 30, 40, 50])
b = np.arange(4)

print("Soustraction", a - b)
print('Puissance :', b**2)
print("Masque booléen (trés utilisé en Pandas) :", a < 35)

# Multiplication matricielle
A = np.array([[1, 1], [0, 1]])
B = np.array([[2, 0], [3, 4]])

print("\nproduit élément par élément (*) :\n", A*B)  # Produit de Hadamard
print("Produit matriciel (@) :\n", A@B, "\nOu\n", A.dot(B))
```

## 5. Agrégations et Upcasting

```python
# Opérations unaires (somme, min, max, etc.)
data = np.random.random((2, 3))
print(data)
print("Somme totale :", data.sum())
print("Minimum par colonne (axis=0) :", data.min(axis=0))
print("Minimum par ligne (axis=1) :", data.min(axis=1))
print("Somme cumulée par ligne (axis=1) :\n", data.cumsum(axis=1))

# Upcasting : NumPy transforme automatiquement le type vers le plus précis
a = np.ones(3, dtype=np.int32)
b = np.linspace(0, np.pi, 3) # float64 par défaut
c = a + b
print(f"\nType de c (int + float) : {c.dtype}")
```

## 6. Opérations "In-place" (Mémoire)
# Utile pour modifier un tableau sans en créer un nouveau (gain de performance)

```python
a = np.ones((2, 3), dtype=int)
b = np.random.random((2, 3))

a *= 3  # Équivalent à a = a * 3, mais plus efficace en mémoire
print("In-place multiplication (a *= 3) :\n", a)

# Attention : l'opération += avec un float sur un array int provoquera une erreur
# car NumPy ne veut pas perdre de précision sans que tu le demandes.
b += a  # OK : b est float, a est int -> b reste float
print("Type de b", b.dtype)
```

## 7. Indexation Booléenne (Le filtre)

## Indexation, slicing et itération

```python
def f(x, y):
    return 10 * x + y

b = np.fromfunction(f, (5, 4), dtype=int)
print("Array complet:\n", b)

# 1. Accès précis [ligne, colonne]
print("\nDeuxième ligne, troisième colonne (b[2, 3]) :", b[2, 3])

# 2. Slicing : [lignes_début:fin , colonnes_début:fin]
print("\nDeuxième colonne pour toutes les lignes (b[:, 1]) :\n", b[:, 1])
print("Lignes 1 à 3 (exclue), toutes les colonnes (b[1:3, :]) :\n", b[1:3, :])

# 3. Accès par la fin
print("\nDernière ligne (b[-1]) :", b[-1])
```

## 9. Itération
Le slicing crée une vue, si on veux un tableau indépendant utilisons `.copy()`

```python
# Itérer sur un tableau multidimensionnel se fait par rapport au premier axe (les lignes)
print("Itération sur les lignes :")
for row in b[:2]: # On limite aux 2 premières pour l'exemple
    print("Ligne :", row)

print("Itération sur les lignes à partir d'un tableau indépendant de b :")
for row in b[:2].copy(): # On slice d'abord, puis on copie
    print("Ligne :", row)

# Si on veut itérer sur chaque élément individuellement : l'attribut .flat
print("\nItération sur chaque élément (b.flat) :")
for element in b.flat[:5]: # 5 premiers éléments
    print(element, end=" ")
```

## 10. L'ellipse (...) : l'outil des fainéants intelligents
# Très pratique pour les tableaux à haute dimension (ex: Images RGB (H, W, C))

```python
c = np.arange(24).reshape(2, 3, 4) # 3 dimensions
print("Shape de c :", c.shape)

# Les deux suivants sont équivalents :
print("\nAccès c[1, :, :] :\n", c[1, :, :])
print("Accès avec ellipse c[1, ...] :\n", c[1, ...])

# Utile pour dire "Prendre tout sur les premières dimensions, et seulement l'index 0 de la dernière"
print("\nDernière dimension index 0 (c[..., 0]) :\n", c[..., 0])
```
