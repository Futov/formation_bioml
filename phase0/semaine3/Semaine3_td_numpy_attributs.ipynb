## Exercice 1 : L'espace et les dimensions (`shape`, `ndim`, `size`)
Nous allons simuler les données de ventes de 3 magasins sur 4 jours.

**Mission :**
1. Crée le tableau.
2. Affiche le nombre de dimensions, la forme exacte du tableau, et le nombre total d'éléments.
3. Essaie d'ajouter un bonus de ventes (un tableau de 4 éléments) et observe ce qui se passe.

```python
# 1. Création du tableau de ventes (3 magasins, 4 jours)
ventes = np.array([
    [120, 130, 110, 150],  # Magasin A
    [90, 85, 100, 110],    # Magasin B
    [200, 210, 195, 220]   # Magasin C
])

# 2. Inspection des attributs
print(f"Dimension (ndim) : {ventes.ndim}D")
line, column = ventes.shape
print(f"Forme (shape) : {ventes.shape}, ({line} lignes, {column} colonnes)")
print(f"Total des éléments (size) : {ventes.size}")

# 3. Le piège du broadcasting
bonus = np.array([10, 20, 15, 30])

# Essaye d'exécuter la ligne suivante.
# Si elle fonctionne, demande-toi pourquoi les dimensions (3,4) et (4,) sont compatibles !
ventes_avec_bonus = ventes + bonus
print("\n ventes avec bonus : \n", ventes_avec_bonus)
```
