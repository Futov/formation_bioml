````python
import csv

patients = [
    ["nom", "age", "sexe", "glycémie", "tension"],
    ["Dupont Jean", 45, "M", 5.2, 120],
    ["Martin Claire", 62, "F", 7.8, 145],
    ["Nguyen Paul", 31, "M", 4.9, 118],
    ["Diallo Fatou", 55, "F", 6.3, 132],
    ["Mbeki Samuel", 70, "M", 8.1, 158],
    ["Leclerc Anne", 28, "F", 4.7, 110],
    ["Koné Ibrahim", 48, "M", 5.9, 128],
    ["Bernard Lucie", 67, "F", 7.2, 142],
]

with open("patients_files/patients_exo.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(patients)

print("Fichier créé.")
````

````python
import csv

# 1. Lire le CSV et retourner une liste de dictionnaires
def lire_patients(file_path):
    """
    Lecture d'un fichier csv et renvoie une liste de dictionnaire
    :param file_path:
    :return:
    """
    with open(file_path, "r", encoding="utf-8") as f:
        lignes = csv.DictReader(f)
        return [ligne for ligne in lignes]

# 2. Filtrer les patients au-dessus d'un seuil d'âge
def filtrer_par_age(patients, age_minimum):
    """
    Filtrer les patients au-dessus du seuil d'âge 'age_minimum'
    :param patients:
    :param age_minimum:
    :return:
    """
    patients_au_dessus_de_age = []
    for ligne in patients:
            if int(ligne["age"]) >= age_minimum:
                patients_au_dessus_de_age.append(ligne)

    return patients_au_dessus_de_age

# 3. Afficher un résumé lisible
def afficher_resume(patients):
    """
    Afficher un résumé lisible de l'état de santé du patient.
    :param patients:
    :return:
    """
    for patient in patients:
        nom = patient['nom']
        age = patient['age']
        glycemie = float(patient['glycémie'])
        # Évaluation de l'état de la glycémie (> 6.5, hyperglycémie)
        if glycemie > 6.5 :
            etat_glycemie = "Hyperglycémie"
        else:
            etat_glycemie = "Glycémie normale"

        print(f"{nom} - {age} ans - {etat_glycemie} ({glycemie})")


# Utilisation attendue
tous_les_patients = lire_patients("patients_files/patients_exo.csv")
#print(tous_les_patients)
patients_seniors = filtrer_par_age(tous_les_patients, age_minimum=50)
#print(patients_seniors)
afficher_resume(patients_seniors)
#```
#
#Le résultat attendu ressemblera à :
#```
#Martin Claire — 62 ans — ⚠ Hyperglycémie (7.8)
#Diallo Fatou  — 55 ans — Glycémie normale (6.3)
#Mbeki Samuel  — 70 ans — ⚠ Hyperglycémie (8.1)
#Bernard Lucie — 67 ans — ⚠ Hyperglycémie (7.2)
````
