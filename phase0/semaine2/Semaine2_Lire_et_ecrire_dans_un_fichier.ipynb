### Les différents modes d'ouverture

```python
# Le mode 'x' dit Création Exclusive
# À utiliser lorsqu'on génère des rapports ou des sauvegardes (Cela évite de détruire des fichiers existant par accident.).
try:
    with open("patients_files/donnees_critiques.txt", "x", encoding="utf-8") as f:  # À préfere à "w"
        f.write("Nouvelles données sécurisées")
        print("Écriture effectuée.")
except FileExistsError:
    print("Erreur : Le fichier existe déjà, opération annulée pour protéger les données.")
```

```python
# Le mode "+" qui permet une lecture/écriture simultanée.
# Il s'ajoute aux modes de base (r+, w+, a+)
# Cependant le mode de base défini le comportement initial.
```

```python
# Le piège du curseur (utilisation de seek)
with open("patients_files/test.txt", "w+", encoding="utf-8") as f:
    f.write("Hello World")

    # Après écriture le fichier est à la fin, alors f.read() ne renverra rien !
    f.seek(0)  # Positionne le curseur au début à l'index 0.
    print(f.read())
```

```python
# Filtrer les patients seniors et exporter
import csv

patients_seniors = []

with open("patients_files/patients.csv", "r", encoding="utf-8") as f:
    lecteur = csv.DictReader(f)
    for ligne in lecteur:
        if int(ligne["age"]) >= 50:
            patients_seniors.append(ligne)

# Exporter dans un nouveau fichier
champ = ["id", "nom", "prenom", "age", "sexe", "glycemie", "imc", "diagnostic"]

with open("patients_files/patients_seniors.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=champ, extrasaction="ignore")
    writer.writeheader()
    writer.writerows(patients_seniors)

print(f"{len(patients_seniors)} patients seniors exportés")
```

```python
# Filtrer les patients dont la glycémie dépasse 6.5 mmol/L (seuil de prédiabète selon l'OMS).
import csv

patients_a_risque = []

with open("patients_files/patients.csv", "r", encoding="utf-8") as f:
    lecteur = csv.DictReader(f)
    for ligne in lecteur:
        if float(ligne["glycemie"]) > 6.5:
            patients_a_risque.append(ligne)

# Champs du fichier exportation
champ = ["id", "nom", "prenom", "age", "sexe", "glycemie", "imc", "diagnostic"]

with open("patients_files/patients_a_risque.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=champ, extrasaction="ignore")
    writer.writeheader()
    writer.writerows(patients_a_risque)

print(f"{len(patients_a_risque)} patients présentant des risque de diabète.")
```
