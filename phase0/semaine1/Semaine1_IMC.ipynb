# Calculer de l'indice de masse corporel (**IMC**)

```python
# Liste des patients.
patients = [
    {"nom": "Patient A", "poids": 70, "taille": 1.75},
    {"nom": "Patient B", "poids": 95, "taille": 1.68},
    {"nom": "Patient C", "poids": 52, "taille": 1.72},
    {"nom": "Patient D", "poids": 110, "taille": 1.80},
    {"nom": "Patient E", "poids": 48, "taille": 1.65},
]
```

```python
def calcul_imc(poids, taille):
  """
  Calcul l'IMC d'un patient.
  """
  return poids / taille**2

def categoriser_imc(imc):
  """
  Interprétation de la valeur de l'IMC.
  """
  if imc < 18.5 :
    return "Sous-poids"
  elif imc < 25:
    return "Poids normal"
  elif imc < 30:
    return "Surpoids"
  else:
    return "Obésité"
```

```python
for patient in patients:
  imc = calcul_imc(patient["poids"], patient["taille"])
  print(f"{patient['nom']} - IMC : {imc:.2f} - {categoriser_imc(imc)}")
```

###Fonction de préparation à l'analyse

```python
def analyser_patient(patient):
  """
  Retourne un dictionnaire contenant le nom, l'IMC et la catégorie du patient.
  """
  imc = calcul_imc(patient["poids"], patient["taille"])
  return {
      "nom": patient["nom"],
      "imc": round(imc,2),
      "categorie": categoriser_imc(imc)
  }
```

```python
# Préparation des résultat à une analyse statistique
resultats = [analyser_patient(patient) for patient in patients]

# Affichage de vérification.
for r in resultats:
  print(f"{r['nom']} - IMC : {r['imc']} - {r['categorie']}")
```
