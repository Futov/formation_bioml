**Définition des fonctions**

```python
def least_difference(a, b, c):
  """
  Return the smallest difference between any two numbers
  among a, b and c.

  >>> least_difference(1, 5, -5)
  4
  """
  diff1 = abs(a - b)
  diff2 = abs(b - c)
  diff3 = abs(a - c)
  return min(diff1, diff2, diff3)
```

```python
print(
    least_difference(1, 10, 100),
    least_difference(1, 10, 10),
    least_difference(5, 6, 7), sep=' result= '
)
```

```python
help(least_difference)
```

### Explication du code :

`print("Hello, World ! Python is amazing programming language", sep='*')`

Cette ligne de code utilise la fonction `print()` de Python pour afficher du texte sur la console.

*   `"Hello, World ! Python is amazing programming language"` : C'est la chaîne de caractères (le texte) que vous souhaitez afficher. Elle est placée entre guillemets doubles.
*   `sep='*'` : C'est un argument optionnel de la fonction `print()`. L'argument `sep` (pour 'separator') spécifie la chaîne de caractères à utiliser comme séparateur entre les différents éléments que vous passez à `print()`. Dans ce cas, comme il n'y a qu'un seul élément (la chaîne de caractères), l'effet du `sep='*'` ne sera pas visible. Si vous aviez plusieurs éléments, par exemple `print("Hello", "World", sep='*')`, le résultat serait `Hello*World`.
